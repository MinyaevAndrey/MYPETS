```
import numpy as np
import pandas as pd
```

```
import kagglehub

path = kagglehub.dataset_download("carolzhangdc/imdb-5000-movie-dataset")
movie = pd.read_csv(path + '/' + 'movie_metadata.csv')
movie.head()
```

**Step 3**

```
genres = movie[['movie_title', 'genres']]
```

```
genres
```

```
def reversator(value):
    return value[::-1]
```

```
for i, row in genres.iterrows():
    print(row)
```

```
# map метод Серии, применяет функцию к каждому ее элементу
for i, row in genres.iterrows():
    print(row.map(reversator))
```

```
for i, col in genres.items():
    print(col)
```

```
for i, col in genres.items():
    print(col.map(reversator))
```

**Step 4**

```
budget=movie[['budget', 'duration']]
```

```
budget.head()
```

```
# применяет ф-цию к каждому значению датасета отдельно
budget.map(lambda x: x + 1).head()
```

```
# применяет функцию к:
# - колонке (axis=0)
# - строке (axis=1)
budget.apply(np.mean, axis=0)
```

```
budget.apply(lambda x: x + 1, axis=0).head()
```

```
# нельзя передать агрегирующую функцию
budget.transform(lambda x: x + 1)
```

```
budget.mean() + 1
```

**Step 5**

```
# np array выигрывает в скорости по сравнению с pd.DataFrame
np.mean(budget['budget'].values)
```

```
# но и имеет свои неудобства: применяем dropna() чтобы посчитать mean()
np.mean(budget['budget'].dropna().values)
```

```
# делаем дату индексом
stock=pd.read_csv('https://raw.githubusercontent.com/PacktPublishing/Pandas-Cookbook/master/data/amzn_stock.csv',
                  index_col='Date',
                  parse_dates=True)
stock.head()
```

```
stock.index
```

```
# фильтр по частичной дате
# stock['2010'] - так не работает
stock['2010':'2010'].head()
```

```
stock['2010-02':'2011-03'].head()
```

```
# все строки за 2010 год
stock.loc['2010'].head()
```

```
# за январь 2010
stock.loc['2010-01'].head()
```

```
stock[stock.index.year == 2010].head()
```

```
# Создает объект-ресемплер, который группирует данные по интервалам в 12 часов.
# Важно: resample() сам по себе ничего не вычисляет — он только подготавливает структуру для последующей агрегации или заполнения.
# Этот метод просто выбирает первую строку из каждой группы (или NaN, если данных нет), без какой-либо агрегации (stock.resample('12h').first()).
stock.resample('12h').asfreq().head()
```

```
stock.resample('12h').first().head()
```

```
stock.resample('1W').mean().head()
```

```
stock.head(10)
```

```
stock.rolling(3)
```

```
stock.rolling(3).mean().head()
```

```
# min_periods=1 позволяет сократить окно расчета, если нет доступных данных
# для 4 числа расчет был только по 4 числу
# для 5 числа - по 4 и 5
# для 6 числа - доступны все данные окна=3
stock.rolling(3, min_periods=1).mean().head()
```

```
# расчет расширяющимся окном, расширение начинается с 1 и до конца
# равные веса для каждого элемента окна
stock.expanding().mean().head()
```

```
# расширениен ачинается с 3
stock.expanding(3).mean().head()
```

```
# расширяющееся окно как expanding только с угасающим весом
stock.ewm(alpha=0.7).mean()
```

```
stock['Open'].plot()
```

```
ns=stock['Open'].rolling(10, min_periods=1).mean()
```

```
stock['Open'].plot()
ns.plot()
```
